# Workbench check: ventas sintéticas

**Pregunta:** ¿Cuál es el promedio de unidades vendidas por día en esta tabla sintética de cinco días?

Selecciona el kernel local `ti2006b-workbench` (Python 3.12), reinícialo y ejecuta todas las celdas. Los datos son inventados y no contienen información personal.


```python
import pandas as pd

ventas = pd.DataFrame({
    "dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "unidades": [12, 15, 9, 18, 16],
})
ventas
```

```python
promedio = ventas["unidades"].mean()
print(f"Promedio diario de ventas (datos sintéticos): {promedio:.2f} unidades")
```

## Interpretación

La tabla suma 70 unidades en cinco días, por lo que el promedio calculado es 14 unidades por día. Este valor resume únicamente los cinco datos sintéticos; no permite inferir ventas reales ni pronosticar otros días. Al ejecutar el notebook, la salida de análisis debe coincidir con este cálculo.
